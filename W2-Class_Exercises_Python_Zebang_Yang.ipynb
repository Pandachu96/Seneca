## 3. Arithmetic Relay
**Goal:** Practice arithmetic operators and precedence (PEMDAS).

1. Compute each expression **by hand** first.
2. Then run the code to verify.


````python
expressions = {
    "4 + 3 * 2": 4 + 3 * 2,
    "(4 + 3) * 2": (4 + 3) * 2,
    "4 ** 2 // 3": 4 ** 2 // 3,
    "22 / 8": 22 / 8,
    "22 // 8": 22 // 8,
    "22 % 8": 22 % 8,
}

for expr, value in expressions.items():
    print(f"{expr} = {value}")
````

## 8. String Formatting Race (`%`, `.format()`, and f-strings)
**Goal:** Format floating-point output.

Given `price = 1.21997`, print:
```
Price per liter: 1.22
```
using all three methods.


````python
price = 1.21997

# Old-style formatting with %
print("Price per liter: %.2f" % price)

# format() method
print("Price per liter: {:.2f}".format(price))

# f-string (Python 3.6+)
print(f"Price per liter: {price:.2f}")
````

## 9. Debugging Challenge: String Immutability
**Goal:** Understand why strings cannot be changed in place.

Why does the following fail? How can we fix it?


````python
# This raises a TypeError because strings are immutable:
# name = "harry"
# name[0] = "H"  # ❌ cannot assign to string index

# Correct approach: create a new string
name = "harry"
fixed = "H" + name[1:]
print(fixed)
````
